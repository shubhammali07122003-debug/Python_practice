# 32. Difference between is and ==

```python
a = [1,2]
b = [1,2]

print(a == b)
print(a is b)
```

```python

a = [1,2,3]
b = [1,2,3]
print(a == b)

# same values
```

```python
a = [1,2,3]
b = a
print(a is b)
# same memory reference
```
